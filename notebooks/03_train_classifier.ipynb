# 🏋️ FakeInversion - Notebook 3: Train Classifier

This notebook:
1. Loads pre-extracted 9-channel features
2. Trains a modified ResNet50 classifier
3. Monitors training with loss/accuracy/AUC curves
4. Saves best model checkpoint

```
# Setup
import os, sys
PROJECT_DIR = '/content/fake_inversion'
os.chdir(PROJECT_DIR)
sys.path.insert(0, PROJECT_DIR)

from google.colab import drive
drive.mount('/content/drive')

import config
config.ensure_dirs()
config.print_config()
```

```
# Check available features
import glob

for split in ['train', 'val', 'test']:
    feat_dir = os.path.join(config.FEATURES_DIR, split)
    n = len(glob.glob(os.path.join(feat_dir, '*.pt'))) if os.path.exists(feat_dir) else 0
    print(f'{split:10s}: {n} features')
```

```
# Build model
from classifier.model import build_model

model = build_model(pretrained=True)
print(model)
```

```
# Train!
from classifier.train import train

# Training with default hyperparameters
model, best_auc = train(
    num_epochs=20,
    learning_rate=1e-4,
    batch_size=32,
)

print(f'\nBest Validation AUC: {best_auc:.4f}')
```

```
# Plot training curves
from evaluation.visualize import plot_training_curves
import matplotlib.pyplot as plt

plot_training_curves(
    save_path=os.path.join(config.RESULTS_DIR, 'plots', 'training_curves.png')
)
plt.show()
```

```
# Quick test on test set
from evaluation.evaluate import Evaluator

evaluator = Evaluator()
test_dir = os.path.join(config.FEATURES_DIR, 'test')
if os.path.exists(test_dir):
    test_metrics = evaluator.evaluate_overall(test_dir)
    print(f"\nTest Accuracy: {test_metrics['accuracy']:.4f}")
    print(f"Test AUC:      {test_metrics['auc']:.4f}")
```
